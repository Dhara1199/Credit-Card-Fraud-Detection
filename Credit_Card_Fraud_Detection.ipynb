# Credit Card Fraud Detection Using Machine Learning
**Aim:** Predict whether a credit card transaction is genuine (0) or fraudulent (1).
Download `creditcard.csv` from the Kaggle link in README and place it beside this notebook.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, average_precision_score
import joblib
```

```python
data = pd.read_csv('creditcard.csv')
print(data.head())
print('Dataset shape:', data.shape)
print(data['Class'].value_counts())
```

```python
sns.countplot(x='Class', data=data)
plt.title('Genuine (0) vs Fraud (1) Transactions')
plt.show()
```

```python
X = data.drop('Class', axis=1)
y = data['Class']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
```

```python
model = RandomForestClassifier(
    n_estimators=150, class_weight='balanced_subsample',
    random_state=42, n_jobs=-1, min_samples_leaf=2
)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
probabilities = model.predict_proba(X_test)[:, 1]
```

```python
print(classification_report(y_test, predictions, target_names=['Genuine', 'Fraud'], zero_division=0))
print('ROC-AUC:', roc_auc_score(y_test, probabilities))
print('PR-AUC:', average_precision_score(y_test, probabilities))
sns.heatmap(confusion_matrix(y_test, predictions), annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
joblib.dump({'model': model, 'feature_names': list(X.columns)}, 'fraud_detection_model.pkl')
print('Model saved successfully.')
```
